# Descriptive Statistics Primer — Titanic Dataset

This notebook calculates and interprets mean, median, mode, standard deviation, and percentiles for numeric columns in the Titanic dataset.

## Cell 1 — Import libraries

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from IPython.display import display
```

## Cell 2 — Load the Titanic dataset

Upload `Titanic-Dataset.csv` to the Colab session, or change the filename below to match your file.

```python
df = pd.read_csv("Titanic-Dataset.csv")
df.head()
```

## Cell 3 — Inspect the dataset

```python
print("Dataset shape:", df.shape)
print("\nColumn names:")
print(df.columns.tolist())
print("\nData types and non-null counts:")
df.info()
print("\nMissing values:")
display(df.isnull().sum().to_frame("Missing Values"))
```

## Cell 4 — Select numeric columns

```python
numeric_columns = ["Age", "Fare", "SibSp", "Parch", "Pclass"]
available_columns = [column for column in numeric_columns if column in df.columns]
numeric_df = df[available_columns].copy()
numeric_df.head()
```

## Cell 5 — Check missing values

```python
missing_summary = numeric_df.isnull().sum().to_frame("Missing Values")
missing_summary["Missing Percentage"] = (numeric_df.isnull().mean() * 100).round(2)
display(missing_summary)
```

## Cell 6 — Calculate summary statistics

```python
summary = pd.DataFrame(index=available_columns)
summary["Count"] = numeric_df.count()
summary["Mean"] = numeric_df.mean()
summary["Median"] = numeric_df.median()
summary["Mode"] = numeric_df.mode().iloc[0]
summary["Standard Deviation"] = numeric_df.std()
summary["Minimum"] = numeric_df.min()
summary["25th Percentile"] = numeric_df.quantile(0.25)
summary["50th Percentile"] = numeric_df.quantile(0.50)
summary["75th Percentile"] = numeric_df.quantile(0.75)
summary["Maximum"] = numeric_df.max()
summary = summary.round(3)
display(summary)
```

## Cell 7 — Calculate the middle 50% range

```python
quartiles = pd.DataFrame(index=available_columns)
quartiles["Q1 (25th Percentile)"] = numeric_df.quantile(0.25)
quartiles["Q3 (75th Percentile)"] = numeric_df.quantile(0.75)
quartiles["IQR (Q3 - Q1)"] = quartiles["Q3 (75th Percentile)"] - quartiles["Q1 (25th Percentile)"]
display(quartiles.round(3))
```

## Cell 8 — Compare mean and median to spot possible skew

```python
skew_check = pd.DataFrame(index=available_columns)
skew_check["Mean"] = numeric_df.mean()
skew_check["Median"] = numeric_df.median()
skew_check["Mean - Median"] = skew_check["Mean"] - skew_check["Median"]
skew_check["Skewness"] = numeric_df.skew()
skew_check["Interpretation"] = np.where(
    skew_check["Skewness"] > 0.5, "Positively skewed",
    np.where(skew_check["Skewness"] < -0.5, "Negatively skewed", "Approximately symmetric")
)
display(skew_check.round(3))
```

## Cell 9 — Plot each numeric distribution

```python
for column in available_columns:
    values = numeric_df[column].dropna()
    plt.figure(figsize=(8, 4))
    plt.hist(values, bins=25, edgecolor="black")
    plt.title(f"Distribution of {column}")
    plt.xlabel(column)
    plt.ylabel("Frequency")
    plt.tight_layout()
    plt.show()
```

## Cell 10 — Compare mean, median and spread

```python
for column in available_columns:
    values = numeric_df[column].dropna()
    plt.figure(figsize=(8, 3))
    plt.boxplot(values, vert=False)
    plt.axvline(values.mean(), linestyle="--", label=f"Mean: {values.mean():.2f}")
    plt.axvline(values.median(), linestyle="-", label=f"Median: {values.median():.2f}")
    plt.title(f"Mean, Median and Spread of {column}")
    plt.xlabel(column)
    plt.legend()
    plt.tight_layout()
    plt.show()
```

## Cell 11 — Print a short interpretation for each column

```python
for column in available_columns:
    values = numeric_df[column].dropna()
    mean_value = values.mean()
    median_value = values.median()
    std_value = values.std()
    q1 = values.quantile(0.25)
    q3 = values.quantile(0.75)
    skew_value = values.skew()

    if skew_value > 0.5:
        skew_text = "positively skewed, with a longer tail toward higher values"
    elif skew_value < -0.5:
        skew_text = "negatively skewed, with a longer tail toward lower values"
    else:
        skew_text = "relatively balanced compared with strongly skewed distributions"

    print(f"{column}: mean={mean_value:.2f}, median={median_value:.2f}, standard deviation={std_value:.2f}.")
    print(f"Middle 50%: {q1:.2f} to {q3:.2f}. The distribution is {skew_text}.\n")
```

## Cell 12 — Export the summary table

```python
summary.to_csv("titanic_descriptive_statistics_summary.csv", index_label="Column")
print("Saved: titanic_descriptive_statistics_summary.csv")
```
